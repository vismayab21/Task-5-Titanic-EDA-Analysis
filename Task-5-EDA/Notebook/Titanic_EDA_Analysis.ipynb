```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set visualization style
sns.set_style("whitegrid")

# Display graphs inside notebook
%matplotlib inline

print("Libraries imported successfully!")
```

```python
# Load Titanic dataset
url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/titanic.csv"

df = pd.read_csv(url)

# Display first five rows
df.head()
```

```python
# Check dataset shape
print("Dataset Shape:", df.shape)

# Check column names
print("\nColumn Names:")
print(df.columns.tolist())

# Check first five rows
df.head()
```

```python
# Dataset information
df.info()
```

```python
# Statistical summary
df.describe()
```

```python
# Count male and female passengers
print(df["sex"].value_counts())

# Count passenger classes
print(df["pclass"].value_counts())

# Count survived and not survived
print(df["survived"].value_counts())
```

```python
# Check missing values
print("Missing Values:")
print(df.isnull().sum())
```

```python
plt.figure(figsize=(12, 6))

sns.heatmap(df.isnull(), cbar=False, cmap="viridis")

plt.title("Missing Values in Titanic Dataset")
plt.show()
```

```python
# Create a copy of original dataset
df_clean = df.copy()

# Fill missing age with median
df_clean["age"] = df_clean["age"].fillna(
    df_clean["age"].median()
)

# Fill missing embark town with mode
df_clean["embark_town"] = df_clean["embark_town"].fillna(
    df_clean["embark_town"].mode()[0]
)

# Fill missing embarked values with mode
df_clean["embarked"] = df_clean["embarked"].fillna(
    df_clean["embarked"].mode()[0]
)

print("Missing values handled successfully!")
print(df_clean.isnull().sum())
```

```python
plt.figure(figsize=(7, 5))

sns.countplot(data=df, x="survived")

plt.title("Survival Count of Titanic Passengers")
plt.xlabel("Survived (0 = No, 1 = Yes)")
plt.ylabel("Number of Passengers")

plt.show()
```

```python
plt.figure(figsize=(7, 5))

sns.countplot(data=df, x="sex")

plt.title("Gender Distribution")
plt.xlabel("Gender")
plt.ylabel("Number of Passengers")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.histplot(df_clean["age"], bins=30, kde=True)

plt.title("Age Distribution of Passengers")
plt.xlabel("Age")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.figure(figsize=(7, 5))

sns.countplot(data=df, x="sex", hue="survived")

plt.title("Survival Based on Gender")
plt.xlabel("Gender")
plt.ylabel("Passenger Count")

plt.legend(title="Survived", labels=["No", "Yes"])

plt.show()
```

```python
plt.figure(figsize=(7, 5))

sns.countplot(data=df, x="pclass", hue="survived")

plt.title("Survival Based on Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Passenger Count")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(data=df_clean, x="survived", y="age")

plt.title("Age Distribution by Survival")
plt.xlabel("Survived (0 = No, 1 = Yes)")
plt.ylabel("Age")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="age",
    y="fare",
    hue="survived",
    alpha=0.7
)

plt.title("Relationship Between Age and Fare")
plt.xlabel("Age")
plt.ylabel("Fare")

plt.show()
```

```python
# Select numerical columns
numeric_df = df_clean.select_dtypes(include="number")

# Calculate correlation
correlation = numeric_df.corr()

plt.figure(figsize=(12, 8))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap of Titanic Dataset")
plt.show()
```

```python
# Select important numerical variables
pair_data = df_clean[
    ["survived", "age", "fare", "pclass", "sibsp", "parch"]
]

sns.pairplot(
    pair_data,
    hue="survived",
    diag_kind="hist"
)

plt.suptitle("Pairplot of Titanic Dataset", y=1.02)

plt.show()
```

```python
print("TOTAL PASSENGERS:", len(df))

print("\nSURVIVAL COUNTS:")
print(df["survived"].value_counts())

print("\nSURVIVAL RATE:")
print(df["survived"].mean() * 100, "%")

print("\nSURVIVAL RATE BY GENDER:")
print(df.groupby("sex")["survived"].mean() * 100)

print("\nSURVIVAL RATE BY CLASS:")
print(df.groupby("pclass")["survived"].mean() * 100)

print("\nAVERAGE AGE:")
print(df["age"].mean())

print("\nAVERAGE FARE:")
print(df["fare"].mean())
```

```python
print("Skewness of numerical columns:")
print(df_clean.select_dtypes(include="number").skew())
```

```python
df_clean["fare_log"] = np.log1p(df_clean["fare"])

plt.figure(figsize=(8, 5))

sns.histplot(df_clean["fare_log"], kde=True)

plt.title("Fare Distribution After Log Transformation")
plt.xlabel("Log Transformed Fare")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(x=df_clean["fare"])

plt.title("Outlier Detection in Fare")
plt.xlabel("Fare")

plt.show()
```

```python
print("========== EDA SUMMARY ==========")

print("1. Total passengers:", len(df))

print("2. Total survivors:", df["survived"].sum())

print("3. Total non-survivors:",
      (df["survived"] == 0).sum())

print("4. Missing values in original dataset:",
      df.isnull().sum().sum())

print("5. Female survival rate:",
      round(df[df["sex"] == "female"]["survived"].mean() * 100, 2), "%")

print("6. Male survival rate:",
      round(df[df["sex"] == "male"]["survived"].mean() * 100, 2), "%")

print("7. Average passenger age:",
      round(df["age"].mean(), 2))

print("========== ANALYSIS COMPLETED ==========")
```
